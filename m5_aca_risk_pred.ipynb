```python
import pandas as pd
import os
from mysql import connector as cnn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from dotenv import load_dotenv
load_dotenv()
db_pass=os.getenv("DB_PASSWORD")
db=os.getenv("DB")
conn=cnn.connect(
    host='localhost',
    user='root',
    password=db_pass,
    database=db
)
cur=conn.cursor()
cur.execute("USE STUDENT")
df=pd.read_sql("SELECT * FROM ACAD_PERF",con=conn)
df
```

```python
x=df.drop(columns=['ST_ID','ACADEMIC_RISK'])
y=df['ACADEMIC_RISK']
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
x_train
```

```python
y_train
```

```python
#Without scaling data and removing outliers
model1=LinearRegression()
model1.fit(x_train,y_train)
pred1=model1.predict(x_test)
print(pred1.shape,y_test.shape)
print(model1.score(x_test,y_test))
pred1
```

```python
#With scaling and removing outliers
df.describe()
```

```python
df.info()
```

```python
#Scaling
import numpy as np
from sklearn.preprocessing import MinMaxScaler,StandardScaler
model2=LinearRegression()
scale_x=MinMaxScaler()
scale_y=MinMaxScaler()
x_train_scaled=scale_x.fit_transform(x_train)
y_train_scaled=scale_y.fit_transform(np.array(y_train).reshape(-1,1))
x_test_scaled=scale_x.transform(x_test)
y_test_scaled=scale_y.transform(np.array(y_test).reshape(-1,1))
```

```python
#Training
model2.fit(x_train_scaled,y_train_scaled)
pred2=model2.predict(x_test_scaled)
pred2=scale_y.inverse_transform(np.array(pred2).reshape(-1,1))
```

```python
#output
print(model2.score(x_test_scaled,y_test_scaled))
pred2
```

```python
from sklearn.metrics import root_mean_squared_error
print(root_mean_squared_error(y_test,pred1))
```

```python
print(root_mean_squared_error(y_test,pred2))
```
